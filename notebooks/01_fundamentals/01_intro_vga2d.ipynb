# 01 — Introduction to VGA2d

This notebook introduces the fundamentals of Geometric Algebra (GA) using 2D Vector Geometric Algebra (VGA2d). We explore why GA is a compelling alternative to matrix-based linear algebra, and we meet the core building blocks: multivectors, blades, and grades.

## Learning Objectives

- Understand what Geometric Algebra is and why it matters
- Construct multivectors using AMSA
- Identify blades and grades
- Visualize the basis structure of VGA2d

```python
# Setup: import AMSA
import matplotlib.pyplot as plt
import numpy as np

from amsa import Algebra
```

## 1.1 Why Geometric Algebra?

Traditional linear algebra uses vectors and matrices. Geometric Algebra extends this with a unified, coordinate-free framework that:

1. **Unifies multiple products** — geometric, outer, inner all in one algebra
2. **Avoids coordinate artifacts** — operations are basis-independent
3. **Handles rotations naturally** — rotors avoid gimbal lock and quaternion awkwardness
4. **Works in any dimension** — same API for 2D, 3D, nD

As physicist David Hestenes put it:

> *Geometry without algebra is dumb! Algebra without geometry is blind!*

Let's start by creating an algebra.

```python
# Create a 2D Vector Geometric Algebra with Euclidean signature
# VGA2d has signature (1, 1) — two positive-definite dimensions

alg = Algebra.vga2d()
print("Algebra spec:", alg.spec)
print("Dimension:", alg.spec.dimension)
print("Signature:", alg.spec.signature)
print("Blade count:", alg.spec.blade_count)
```

## 1.2 The Basis Blades of VGA2d

VGA2d has **4 basis blades** — one for each grade:

| Grade | Name | Description |
|-------|------|-------------|
| 0 | **e** (scalar) | Magnitude only, no direction |
| 1 | **e1** | Unit vector in x-direction |
| 1 | **e2** | Unit vector in y-direction |
| 2 | **e12** | Unit bivector (oriented area) |

The total dimension is 2² = 4 blades.

```python
# Explore the basis blades
print("All blades in the algebra:")
grades = alg.spec.grades_of_blades()
for blade_index in range(alg.spec.blade_count):
    name = alg.spec.blade_name(blade_index)
    grade = grades[blade_index]
    print(f"  Blade {blade_index}: '{name}' (grade {grade})")
```

## 1.3 Constructing Multivectors

A **multivector** is a linear combination of basis blades. In AMSA, we construct them using various helpers.

```python
# Scalar (grade 0)
s = alg.scalar(3.5)
print("Scalar:", s.values)
print("  layout blades:", s.layout.blades)

# Vector (grade 1) — from coordinates
v = alg.vector([1.0, 2.0])  # 1*e1 + 2*e2
print("\nVector:", v.values)
print("  layout blades:", v.layout.blades)

# Bivector (grade 2) — explicit blade construction
b = alg.multivector({"e12": 1.0})
print("\nBivector:", b.values)
print("  layout blades:", b.layout.blades)

# Arbitrary multivector (mixed grades)
mv = alg.multivector({"e": 1.0, "e1": 2.0, "e2": 3.0, "e12": 4.0})
print("\nMixed multivector:", mv.values)
print("  layout blades:", mv.layout.blades)
```

## 1.4 Grade Projection

We can extract specific grades from a multivector using the `grade()` method.

```python
mv = alg.multivector({"e": 1.0, "e1": 2.0, "e2": 3.0, "e12": 4.0})

# Extract each grade
scalar_part = mv.grade(0)
vector_part = mv.grade(1)
bivector_part = mv.grade(2)

print("Original:", mv.values)
print("Grade 0 (scalar):", scalar_part.values)
print("Grade 1 (vector):", vector_part.values)
print("Grade 2 (bivector):", bivector_part.values)
```

## 1.5 Visualizing the VGA2d Basis

Let's create a visual representation of the VGA2d basis to build intuition.

```python
fig, axes = plt.subplots(1, 4, figsize=(14, 3))

# Scalar — represented as a point at origin
axes[0].scatter([0], [0], s=200, c='black')
axes[0].text(0, 0.3, 'e (scalar)', ha='center', fontsize=12, fontweight='bold')
axes[0].set_xlim(-1, 1)
axes[0].set_ylim(-1, 1)
axes[0].set_title('Grade 0: Scalar', fontsize=11)
axes[0].set_aspect('equal')
axes[0].axis('off')

# Grade 1: e1 (x-axis)
axes[1].arrow(0, 0, 0.7, 0, head_width=0.1, head_length=0.05, fc='red', ec='red', linewidth=2)
axes[1].text(0.4, -0.2, 'e1', ha='center', fontsize=12, fontweight='bold', color='red')
axes[1].set_xlim(-1, 1)
axes[1].set_ylim(-1, 1)
axes[1].set_title('Grade 1: e1', fontsize=11)
axes[1].set_aspect('equal')
axes[1].axis('off')

# Grade 1: e2 (y-axis)
axes[2].arrow(0, 0, 0, 0.7, head_width=0.1, head_length=0.05, fc='blue', ec='blue', linewidth=2)
axes[2].text(-0.2, 0.4, 'e2', ha='center', fontsize=12, fontweight='bold', color='blue')
axes[2].set_xlim(-1, 1)
axes[2].set_ylim(-1, 1)
axes[2].set_title('Grade 1: e2', fontsize=11)
axes[2].set_aspect('equal')
axes[2].axis('off')

# Grade 2: e12 (oriented area — draw a square with curved arrow)
rect = plt.Rectangle((-0.3, -0.3), 0.6, 0.6, fill=False, edgecolor='purple', linewidth=2)
axes[3].add_patch(rect)
# Add curved arrow to indicate orientation
from matplotlib.patches import Arc

arc = Arc((0, 0), 0.4, 0.4, theta1=0, theta2=270, color='purple', linewidth=2)
axes[3].add_patch(arc)
axes[3].text(0, 0, 'e12', ha='center', va='center', fontsize=12, fontweight='bold', color='purple')
axes[3].set_xlim(-1, 1)
axes[3].set_ylim(-1, 1)
axes[3].set_title('Grade 2: e12 (bivector)', fontsize=11)
axes[3].set_aspect('equal')
axes[3].axis('off')

plt.suptitle('VGA2d Basis Blades', fontsize=14, fontweight='bold', y=1.05)
plt.tight_layout()
plt.show()
```

## 1.6 Summary

In this notebook we covered:

- **VGA2d** has 4 basis blades: scalar (e), two vectors (e1, e2), and one bivector (e12)
- **Multivectors** are combinations of these blades
- **Grades** organize blades by their level: 0=scalar, 1=vector, 2=bivector
- **AMSA** provides `scalar()`, `vector()`, `multivector()`, and `grade()` to construct and decompose multivectors
