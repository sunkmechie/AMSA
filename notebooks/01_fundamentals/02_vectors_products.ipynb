# 02 — Vectors and Products

This notebook explores the three fundamental products in Geometric Algebra: geometric, outer, and inner products. We'll see how they decompose the geometric product and build intuition for each.

## Learning Objectives

- Compute the geometric product and understand its decomposition
- Distinguish outer product (wedge) and inner product (contraction)
- Visualize geometric meaning of each product
- Verify the identity: geometric = inner + outer

```python
# Setup
from amsa import Algebra
import numpy as np
import matplotlib.pyplot as plt

alg = Algebra.vga2d()
```

## 2.1 The Geometric Product

The **geometric product** of two vectors produces a multivector:

$$uv = u \cdot v + u \wedge v$$

where:
- **scalar part** ($u \cdot v$) — the inner product, related to projection
- **bivector part** ($u \wedge v$) — the outer product, related to oriented area

This is powerful: one operation captures both projection and area!


```python
# Define two vectors
u = alg.vector([1.0, 2.0])
v = alg.vector([3.0, -4.0])

# Geometric product
gp = u * v  # using the * operator

print("Vector u:", u.values)
print("Vector v:", v.values)
print("\nGeometric product u*v:", gp.values)
print("Layout blades:", gp.layout.blades)
```

## 2.2 Decomposing the Geometric Product

The geometric product splits into scalar (inner) and bivector (outer) parts. Let's extract them.

```python
# Decompose the geometric product
scalar_part = gp.grade(0)
bivector_part = gp.grade(2)

print("Scalar part (inner product):", scalar_part.values)
print("Bivector part (outer product):", bivector_part.values)

# Alternative: use the outer and inner product operators directly
outer = u ^ v    # wedge product
inner = u | v   # inner product (contraction)

print("\nOuter product u ^ v:", outer.values)
print("Inner product u | v:", inner.values)
```

## 2.3 Verification: Geometric = Inner + Outer

A fundamental identity in GA:

$$u \cdot v = u | v + u \wedge v$$

Let's verify this numerically.

```python
# Verify the decomposition
reconstructed = inner + outer

print("Geometric product:", gp.values)
print("Inner + Outer:", reconstructed.values)
print("\nVerification passed:", np.allclose(gp.values, reconstructed.values))
```

## 2.4 Geometric Interpretation

Each product has a clear geometric meaning:

| Product | Meaning | 2D Result |
|---------|---------|------------|
 | **Inner** ($u|v$) | Projection of u onto v, or magnitude × cos(θ) | Scalar |
 | **Outer** ($u \wedge v$) | Oriented area, magnitude × sin(θ) | Bivector |
 | **Geometric** ($u \cdot v$) | Both together | Multivector |

Let's visualize this.

```python
# Visualize the geometric meaning
fig, axes = plt.subplots(1, 3, figsize=(14, 4.5))

# Get vector components
u_vals = u.grade(1).values
v_vals = v.grade(1).values

# Plot vectors
for ax in axes:
    ax.arrow(0, 0, u_vals[0]*0.5, u_vals[1]*0.5, head_width=0.15, head_length=0.1, fc='blue', ec='blue', linewidth=2, label='u')
    ax.arrow(0, 0, v_vals[0]*0.5, v_vals[1]*0.5, head_width=0.15, head_length=0.1, fc='red', ec='red', linewidth=2, label='v')
    ax.set_xlim(-2.5, 2.5)
    ax.set_ylim(-2.5, 2.5)
    ax.set_aspect('equal')
    ax.grid(True, alpha=0.3)
    ax.legend()
    ax.axhline(0, color='gray', linewidth=0.5)
    ax.axvline(0, color='gray', linewidth=0.5)

axes[0].set_title(f'Vectors u and v\nInner (scalar): {inner.values[0]:.2f}', fontsize=11)

# Add outer product visualization
axes[1].fill([0, u_vals[0], u_vals[0]+v_vals[0], v_vals[0]], 
             [0, u_vals[1], u_vals[1]+v_vals[1], v_vals[1]], 
             alpha=0.3, color='purple')
axes[1].set_title(f'Outer (bivector): {outer.values[0]:.2f} * e12\nArea = {abs(outer.values[0]):.2f}', fontsize=11)

# Combined
axes[2].fill([0, u_vals[0], u_vals[0]+v_vals[0], v_vals[0]], 
             [0, u_vals[1], u_vals[1]+v_vals[1], v_vals[1]], 
             alpha=0.3, color='purple')
axes[2].set_title(f'Geometric Product\nScalar: {scalar_part.values[0]:.2f}, Bivector: {bivector_part.values[0]:.2f}', fontsize=11)

plt.tight_layout()
plt.show()
```

## 2.5 Special Cases

Let's explore what happens with perpendicular and parallel vectors.

```python
# Perpendicular vectors
u_perp = alg.vector([1.0, 0.0])
v_perp = alg.vector([0.0, 1.0])

gp_perp = u_perp * v_perp
outer_perp = u_perp ^ v_perp
inner_perp = u_perp | v_perp

print("=== Perpendicular vectors (e1, e2) ===")
print("Geometric:", gp_perp.values)
print("Outer:", outer_perp.values)
print("Inner:", inner_perp.values)
print("\nNote: perpendicular → inner is zero, outer gives max area")
```

```python
# Parallel vectors
u_par = alg.vector([2.0, 0.0])
v_par = alg.vector([3.0, 0.0])

gp_par = u_par * v_par
outer_par = u_par ^ v_par
inner_par = u_par | v_par

print("=== Parallel vectors (both along x) ===")
print("Geometric:", gp_par.values)
print("Outer:", outer_par.values)
print("Inner:", inner_par.values)
print("\nNote: parallel → outer is zero, inner gives magnitude product")
```

```python
# Same direction, opposite orientation
u_opp = alg.vector([2.0, 0.0])
v_opp = alg.vector([-3.0, 0.0])

gp_opp = u_opp * v_opp
outer_opp = u_opp ^ v_opp
inner_opp = u_opp | v_opp

print("=== Opposite direction (2*e1, -3*e1) ===")
print("Geometric:", gp_opp.values)
print("Outer:", outer_opp.values)
print("Inner:", inner_opp.values)
print("\nNote: opposite → negative inner (angle > 90°)")
```

## 2.6 Self Product and Squaring

Multiplying a vector by itself gives a pure scalar:

```python
# Vector squared
v = alg.vector([3.0, 4.0])
v_squared = v * v

print("Vector v:", v.values)
print("v * v:", v_squared.values)
print("\nNote: v*v is pure scalar (the squared norm)")
print("Magnitude squared:", v_squared.grade(0).values[0])
print("Classical check: 3² + 4² =", 3**2 + 4**2)
```

## 2.7 Summary

We covered:

- **Geometric product** `*` decomposes into inner + outer
- **Outer product** `^` gives oriented area (bivector)
- **Inner product** `|` gives projection/scalar
- Perpendicular vectors → zero inner, max outer
- Parallel vectors → zero outer, max inner
- $v \cdot v = |v|^2$ (squared norm)
