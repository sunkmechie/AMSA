# 01 – VGA Rotors

This notebook introduces AMSA through the 2D and 3D Euclidean geometric algebras (VGA).
We will cover vectors, the geometric product, outer and inner products, rotors, and sandwich conjugation.

```python
import numpy as np

from amsa import Algebra
```

## 1.1 Constructing vectors in VGA2d

```python
alg = Algebra.vga2d()
u = alg.vector([1.0, 2.0])
v = alg.vector([3.0, -4.0])

print("u layout:", u.layout.blades)
print("v values:", v.values)
```

## 1.2 Geometric, outer, and inner products

```python
gp = u * v
op = u ^ v
ip = u | v

print("gp =", gp.as_dense().values)
print("op =", op.values)
print("ip =", ip.values)

# Verify decomposition
assert np.allclose(gp.as_dense().values, (ip + op).as_dense().values)
```

## 1.3 Rotors and sandwich conjugation

A rotor in VGA2d is an even-grade element. We can rotate `e1` onto `e2` with a normalized rotor.

```python
rotor = alg.multivector({"e": np.sqrt(0.5), "e12": -np.sqrt(0.5)})
e1 = alg.blade("e1")
rotated = rotor.sandwich(e1)

print("rotated e1 =", rotated.as_dense().values)
print("e1 component:", rotated.component("e1"))
print("e2 component:", rotated.component("e2"))
```

## 1.4 Batched rotors

All operations broadcast over NumPy batch dimensions.

```python
angles = np.array([0.0, np.pi / 4, np.pi / 2])
cos = np.cos(angles / 2)
sin = np.sin(angles / 2)

batch_rotor = alg.even(np.stack([cos, -sin], axis=-1))
batch_rotated = batch_rotor.sandwich(e1)

print("batch shape:", batch_rotated.batch_shape)
print("values:\n", batch_rotated.as_dense().values)
```
