# 02 – PGA Rigid Body Motion

This notebook introduces 2D Projective Geometric Algebra (PGA2d) in AMSA.
We show how lines, points, and motors are represented, and how to perform rigid-body motions.

```python
import numpy as np

from amsa import Algebra
```

## 2.1 Points, lines, and dual encodings in PGA2d

AMSA currently uses two complementary PGA2d encodings:

- geometric line computations use line bivectors `a e01 + b e02 + c e12`
- motor/viz examples use plottable point bivectors `x e01 + y e02 + w e12`
- the raw meet of two lines returns a vector-form point, which we can convert with `poincare_dual()` for plotting

```python
alg = Algebra.pga2d()

# A line: y - 1 = 0 -> 0*e01 + 1*e02 - 1*e12
line = alg.multivector({"e01": 0.0, "e02": 1.0, "e12": -1.0})

# A plottable / motor point: x*e01 + y*e02 + 1*e12
point = alg.multivector({"e01": 1.0, "e02": 2.0, "e12": 1.0})

print("line layout:", line.layout.blades)
print("point layout:", point.layout.blades)
```

## 2.2 Meet (regressive product) and join (outer product)

The meet of two line bivectors gives a vector-form intersection point.
Applying `poincare_dual()` converts that point into the plotting/motor encoding.

```python
x_axis = alg.multivector({"e01": 0.0, "e02": 1.0, "e12": 0.0})
y_axis = alg.multivector({"e01": 1.0, "e02": 0.0, "e12": 0.0})

origin = x_axis.regress(y_axis)
origin_plot = origin.poincare_dual()
print("Meet of x and y axes (vector form) =", origin.as_dense().values)
print("Meet converted for plotting =", origin_plot.as_dense().values)
```

## 2.3 Motors and sandwich motion

A motor is an even-grade element that encodes rotation and translation.
We apply it via sandwich conjugation.

```python
angle = np.pi / 2
motor = alg.multivector({"e": np.cos(angle / 2), "e12": -np.sin(angle / 2)})
motor = motor.normalized()

p = alg.multivector({"e01": 1.0, "e02": 0.0, "e12": 1.0})
p_rotated = motor.sandwich(p)

print("original point =", p.as_dense().values)
print("rotated point =", p_rotated.as_dense().values)
```

## 2.4 Bulk and weight split

PGA has a null basis `e0` with `e0*e0 = 0`. Elements can be split into
**bulk** (no `e0` factor) and **weight** (contains `e0` factor).

```python
mv = alg.multivector({"e1": 3.0, "e01": 4.0})

print("bulk =", mv.bulk().as_dense().values)
print("weight =", mv.weight().as_dense().values)
print("bulk + weight =", (mv.bulk() + mv.weight()).as_dense().values)

print("bulk norm =", mv.bulk_norm().values)
print("weight norm =", mv.weight_norm().values)
print("bulk normalized =", mv.bulk_normalized().as_dense().values)
print("unitized =", mv.unitized().as_dense().values)
```
